```python
import numpy as np
import matplotlib.pyplot as plt

# Загрузка синтетических данных
url = "https://raw.githubusercontent.com/materials-stepik/Numpy/refs/heads/main/sales_data.csv"
raw_data = np.genfromtxt(url)

print("Количество пропусков:", sum(np.isnan(raw_data)))
# Обработка NaN — замена на среднее
mean_temp = np.nanmean(raw_data)
sales = np.where(np.isnan(raw_data), mean_temp, raw_data)

print("Продажи:", sales.shape)
```

```python
# Рассчитываем статистику
mean_sales = np.mean(sales)
std_sales = np.std(sales)
min_sales = np.min(sales)
max_sales = np.max(sales)

# Печатаем результаты
print(f"Среднее значение: {mean_sales}")
print(f"Стандартное отклонение: {std_sales}")
print(f"Минимум: {min_sales}")
print(f"Максимум: {max_sales}")
```

```python
# Рассчитываем разности
sales_diff = np.diff(sales)
print(f"Максимальная разность между днями: {sales_diff.max():.2f}, средняя разность между днями: {sales_diff.mean():.2f}")
```

```python
import matplotlib.pyplot as plt

# Визуализируем данные
plt.plot(sales, label="Продажи")
plt.xlabel('Дни')
plt.ylabel('Объем продаж')
plt.title('Продажи по дням')
plt.legend()
plt.show()
```

```python
sales_diff_200 = sales_diff[sales_diff>200]
sales_diff_200
```

```python
#скользящее среднее по продажам с окном 7 дней
sales = sales

# Размер окна
window_size = 7

# Создаем ядро для скользящего среднего: массив из единиц, деленный на размер окна
kernel = np.ones(window_size) / window_size

# Используем функцию convolve с режимом 'valid' или 'same'
# 'same' возвращает массив той же длины, что и исходные данные
rolling_sales = np.convolve(sales, kernel, mode='same')

print("Скользящее среднее по продажам:")
for day, value in enumerate(rolling_sales, start=1):
    print(f"День {day}: {value:.2f}")
```

```python
# Визуализируем данные
plt.plot(sales, label="Продажи")
plt.plot(rolling_sales, label="Сглаженные данные", color='red')
plt.xlabel('Дни')
plt.ylabel('Объем продаж')
plt.title('Продажи по дням')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Загрузка синтетических данных
url = "https://raw.githubusercontent.com/materials-stepik/Numpy/refs/heads/main/temperature.txt"
raw_data = np.genfromtxt(url)

# Обработка NaN — замена на среднее
mean_temp = np.nanmean(raw_data)
temperatures = np.where(np.isnan(raw_data), mean_temp, raw_data)

print("Температуры:", temperatures)
```

```python
# Рассчитываем статистику
mean_temp = np.mean(temperatures)
std_temp = np.std(temperatures)
min_temp = np.min(temperatures)
max_temp = np.max(temperatures)

# Печатаем результаты
print(f"Среднее значение: {mean_temp}")
print(f"Стандартное отклонение: {std_temp}")
print(f"Минимум: {min_temp}")
print(f"Максимум: {max_temp}")
```

```python
# Рассчитываем разности
temp_diff = np.diff(temperatures)
print(f"Разности между днями: {temp_diff}")
```

```python
import matplotlib.pyplot as plt

# Визуализируем данные
plt.plot(temperatures, label="Температура")
plt.xlabel('Дни')
plt.ylabel('Температура (°C)')
plt.title('Температура по дням')
plt.legend()
plt.show()
```

```python
# Скользящее среднее с размером окна 5
smoothed_temperatures = np.convolve(temperatures, np.ones(5)/5, mode='valid')

# Визуализируем исходные и сглаженные данные
plt.plot(temperatures, label="Исходные данные")
plt.plot(smoothed_temperatures, label="Сглаженные данные", color='red')
plt.xlabel('Дни')
plt.ylabel('Температура (°C)')
plt.title('Сглаживание временных рядов')
plt.legend()
plt.show()
```

```python
def filter_outliers(data, threshold=2):
    median = np.median(data)
    return np.where(np.abs(data - median) > threshold, median, data)

filtered = filter_outliers(temperatures)

plt.plot(temperatures, label="Оригинал")
plt.plot(filtered, label="Фильтрованные", color='green')
plt.title("Фильтрация выбросов")
plt.xlabel("Дни")
plt.ylabel("Температура (°C)")
plt.legend()
plt.show()
```

```python
def autocorrelation(x, lag=1):
    return np.corrcoef(x[:-lag], x[lag:])[0, 1]

for lag in [1, 2, 3]:
    print(f"Лаг {lag}: автокорреляция = {autocorrelation(temperatures, lag):.3f}")
# Автокорреляция помогает выявить зависимости, которые могут быть полезными для прогнозирования. 
# Например, если на лаге 1 автокорреляция высокая, это значит, что текущие значения сильно зависят от предыдущих
```

```python
# Линейная регрессию для выявления общего тренда и прогноз температуры на следующий день
x = np.arange(len(temperatures))
coeffs = np.polyfit(x, temperatures, deg=1)
trend = np.polyval(coeffs, x)

plt.plot(temperatures, label="Оригинал")
plt.plot(trend, label="Тренд", color="orange")
plt.title("Линейный тренд")
plt.xlabel("Дни")
plt.ylabel("Температура (°C)")
plt.legend()
plt.show()

# Прогноз на следующий день
next_value = np.polyval(coeffs, len(temperatures))
print("Прогноз температуры на следующий день:", next_value)
```
