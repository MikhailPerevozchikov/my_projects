```python
chunk_size = 100
chunks = pd.read_csv("C:\\Users\\Миша\\Downloads\\titanic.csv", chunksize=chunk_size)
total_age = 0
total_count = 0
for chunk in chunks:
    total_age += chunk['Age'].sum()
    total_count += chunk['Age'].count()
average_age = total_age / total_count
print(f"Средний возраст всех пассажиров: {average_age.round(1)}")
```

```python
chunk_size = 200
chunks = pd.read_csv("C:\\Users\\Миша\\Downloads\\titanic.csv", chunksize=chunk_size)
fare_max = None
for chunk in chunks:
    current_max = chunk['Fare'].max()
    if fare_max is None or (current_max > fare_max):
        fare_max = current_max
print(f"максимальное значение в колонке Fare по всему датасету: {fare_max}")    
```

КАСТОМНЫЕ ФУНКЦИИ

```python
data = {'Category': ['X', 'Y', 'Y', 'Z'], 'Amount': [50, 60, 70, 80]}
df = pd.DataFrame(data)
grouped = df.groupby('Category')['Amount'].mean()
print(grouped)
```

```python
data = {'A': [1, 2, 3, 4, 5], 'B': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

def range_calc(series):
    return series.max() - series.min()
df['A'].agg(range_calc)
```

```python
data = {'Category': ['A', 'A', 'B', 'B', 'C'], 
        'Values': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

def med_x_2(series):
    return series.mean() * 2

grouped = df.groupby('Category')['Values'].agg(med_x_2)
grouped
```

```python
data = {'A': [1, 2, 3, 4, 5], 'B': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

def normilize(series):
    return (series - series.min()) / (series.max() - series.min())

result = df.apply(normilize)
result
```

```python
data = {'A': [1, 2, None, 4, 5]} 
df = pd.DataFrame(data)

def sum_skipna(series):
    return series.sum(skipna=True)

df['A'].agg(sum_skipna)
```

```python
data = {'Category': ['X', 'X', 'Y', 'Y', 'Z'], 'Amount': [5, 10, 15, 20, 25]}
df = pd.DataFrame(data)

def custom_med(series):
    return series.mean() * 2

grouped = df.groupby('Category')['Amount'].agg(custom_med)
grouped
```

```python
data = {'X': [2, 4, 6], 'Y': [1, 3, 5]}
df = pd.DataFrame(data)

def custom(series):
    return series.max() - series.min()

df.apply(custom)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({'Scores': [65, 70, 75, 80, 85, 90, 95]})
data['Scores'].plot(kind='hist', bins=5, color='skyblue', edgecolor='black')
plt.title('Распределение оценок')
plt.xlabel('Оценки')
plt.ylabel('Частота')
plt.show()
```

```python
data.boxplot(column='Scores', grid=False, color='blue')
plt.title('Диаграмма размаха оценок')
plt.ylabel('Оценки')
plt.show()
```

```python
data = pd.DataFrame({'Math': [60, 70, 80, 90, 85], 'Science': [55, 65, 75, 85, 80]})
data.plot(kind='scatter', x='Math', y='Science', color='green')
plt.title('Связь между математикой и наукой')
plt.xlabel('Математика')
plt.ylabel('Наука')
plt.show()
```
