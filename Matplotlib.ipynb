```python
import numpy as np
import matplotlib.pyplot as plt

# Создаем сетку координат
x = np.linspace(-6, 6, 100)
y = np.linspace(-6, 6, 100)
X, Y = np.meshgrid(x, y)

# Вычисляем значения функции
Z = np.sin(np.sqrt(X**2 + Y**2))

# Строим контурный график
plt.contour(X, Y, Z, levels=15, colors='blue')

plt.title('Контурный график функции sin(sqrt(x² + y²))')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()
```

```python
import numpy as np
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# 1. Создаем синтетический набор данных moons
X, _ = make_moons(n_samples=500, noise=0.1, random_state=42)

# 2. Подготавливаем координаты сетки
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

xx, yy = np.mgrid[x_min:x_max:.01, y_min:y_max:.01]
positions = np.vstack([xx.ravel(), yy.ravel()])

# 3. Вычисляем оценку плотности методом Гауссового ядра
kernel = gaussian_kde(X.T)
Z = np.reshape(kernel(positions).T, xx.shape)

# 4. Контурная карта плотности
fig, ax = plt.subplots(figsize=(8, 6))
contour_plot = ax.contourf(xx, yy, Z, levels=20, cmap='Blues')
scatter_points = ax.scatter(X[:, 0], X[:, 1], marker='o', color='darkred', s=20, alpha=0.5)

# Оформляем график
ax.set_title('Оценка плотности ядер для datasets.make_moons')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')

# Добавляем colorbar
colorbar = fig.colorbar(contour_plot)
colorbar.set_label('Плотность точек')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Создаем сетку координат
X = np.arange(-2, 2.05, 0.05)
Y = np.arange(-2, 2.05, 0.05)
XX, YY = np.meshgrid(X, Y)

# 2. Вычисляем значения функции Z
ZZ = np.exp(-(XX**2 + YY**2))

# 3. Рисуем контурный график с заливкой цветом
plt.figure(figsize=(8, 6))  # устанавливаем размер фигуры
CSF = plt.contourf(XX, YY, ZZ, levels=30, cmap='viridis')  # 30 уровней и палитру viridis

# 4. Добавляем цветовую шкалу
cb = plt.colorbar(CSF)
cb.set_label('$Z$', rotation=0)  # подпись к цветной шкале

# 5. Обозначаем оси и добавляем название графика
plt.xlabel('$X$')
plt.ylabel('$Y$')
plt.title('Контурный график функции $Z = e^{-(X^2 + Y^2)}$')

# 6. Выводим итоговый график
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Определяем диапазон и шаг
x = np.arange(-3, 3.1, 0.1)
y = np.arange(-3, 3.1, 0.1)

# 2. Создаем сетку координат
X, Y = np.meshgrid(x, y)

# 3. Вычисляем значения функции Z
Z = np.sin(X**2 + Y**2)

# 4. Строим контурный график
plt.figure(figsize=(8, 6))
cs = plt.contour(X, Y, Z, 20, colors='k')  # Число уровней 20, чёрные линии
plt.clabel(cs, inline=1, fontsize=10)       # Подписываем некоторые уровни

# 5. Добавляем сетку и подписи осей
plt.grid(True)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Контурный график функции $\\sin(X^2 + Y^2)$')

# 6. Показываем график
plt.show()
```
