```python
# Загружаем датасет
df = sns.load_dataset("penguins")

# Удаляем строки с пропущенными значениями
df = df.dropna()

# Строим матрицу корреляций
correlation_matrix = df.corr(numeric_only=True)

# Показываем результат
print(correlation_matrix)
```

```python
import numpy as np
# Извлекаем нижнюю треугольную матрицу (без диагонали)
lower_triangular = correlation_matrix.where(np.tril(np.ones(correlation_matrix.shape)).astype(bool))

# Находим минимальную отрицательную корреляцию
min_corr_value = lower_triangular.min().min()

# Получаем индексы пары признаков с минимальной корреляцией
pair = lower_triangular.stack().idxmin()

# Распаковываем имена признаков
feature_1, feature_2 = pair

# Печать результата
print(f"Пара признаков с самой сильной отрицательной корреляцией:\n'{feature_1}' и '{feature_2}',\nЗначение корреляции: {min_corr_value:.2f}")
```

```python
# Извлекаем строку корреляций с flipper_length_mm
correlations_with_flipper_length = correlation_matrix["flipper_length_mm"]

# Сортируем переменные по модулю корреляции с flipper_length_mm
sorted_correlations = correlations_with_flipper_length.abs().sort_values(ascending=False)

# Выводим результат
print(sorted_correlations)
```

```python
from sklearn.datasets import load_diabetes
import pandas as pd

# Загружаем данные
data = load_diabetes(as_frame=True)
df = data.frame

# Посмотрим на структуру данных
df.head()
```

```python
df.info()
```

```python
# Выбираем признаки (X) и целевую переменную (y)
X = df[['age', 'sex', 'bmi', 'bp']]  # возраст, пол, индекс массы тела, артериальное давление
y = df['target']  # целевая переменная
```

```python
# Разделение на обучающую и тестовую выборки
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression

# Создаём и обучаем модель
model = LinearRegression()
model.fit(X_train, y_train)
```

Оценка качества модели
Предскажем значения и рассчитаем метрики:

MAE (средняя абсолютная ошибка) — насколько в среднем модель ошибается.

RMSE (корень из средней квадратичной ошибки) — чувствительная к большим отклонениям.

R² (коэффициент детерминации) — насколько хорошо модель объясняет данные (от 0 до 1).

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f'MAE: {mae:.3f}')
print(f'RMSE: {rmse:.3f}')
print(f'R²: {r2:.3f}')
```

Интерпретация коэффициентов
Каждый коэффициент в модели отражает, насколько целевая переменная изменится, если увеличить признак на 1 единицу (при прочих равных условиях):

```python
coefficients = pd.DataFrame({
    'Признак': X.columns,
    'Коэффициент': model.coef_
})
print(coefficients)
```
